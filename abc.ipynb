```python
import pandas as pd
import numpy as np
```

```python
dct={}
```

```python
def sub(a):

    a=a.replace("-"," ")
    a=a.replace(":"," ")
    x=a.split()
    for i in range(len(x)):
        x[i]=int(x[i])
    return x
```

```python
def fun(userid,time,log_ty):
    if log_ty == "login":
        sup={log_ty:time}
        if userid in dct and len(dct[userid][-1])==2:
            dct[userid].append(sup)
        elif userid in dct and len(dct[userid][-1])==1:
            return "you can login two times without logout"
        else:
            dct[userid]=[sup]
    else:
        if userid not in dct or "logout" in dct[userid][-1]:
            return "user have to login before logout"
        elif True:
            z1=np.array(sub(dct[userid][-1]["login"]))
            z2=np.array(sub(time))
            for i in range(0,len(z1)):
                if z1[i]>z2[i]:
                    dct[userid][-1][log_ty]=time
                    break
                if z1[i]>z2[i]:
                    return "logout time can not be smaller than login time"
                    break
            else:
                dct[userid][-1][log_ty]=time
```

```python
def time_spent():
    year=365*24*60*60
    month=30*24*60*60
    day=24*60*60
    users={}
    for i,j in dct.items():
        l=np.array([0,0,0,0,0,0])
        for k in j:
            if "logout" in k:
                ko=np.array(sub(k["logout"]))
                kl=np.array(sub(k["login"]))
                l=(ko-kl)+l

        users[i]=l

    for i,j in users.items():
        s=sum([j[0]*year,j[1]*month,j[2]*day,j[3]*3600,j[4]*60,j[5]])
        if s<0:
            print(i,":","logout time should be greater than login time")
        else:
            print(i,":",s,"sec")
```

```python
def put_data(x):
    data= pd.read_csv(x)
    data["Timestamp"]=pd.to_datetime(data["Timestamp"])
    for i in data["UserId"].unique():
        user=data[data["UserId"]==i]
        user=user.sort_values(by="Timestamp")
        for j in range(len(user)):
            fun(i,str(user["Timestamp"].iloc[j]),str(user["Log_type"].iloc[j]))
    time_spent()
    
```
